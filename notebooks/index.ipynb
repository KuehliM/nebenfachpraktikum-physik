# Nebenfachpraktikum Physik — Übersicht

Willkommen! Dieses JupyterLab-Profil wird bei jedem Login automatisch mit dem [Praktikums-Repository](.) synchronisiert (`nbgitpuller`). Eure eigenen Änderungen an bestehenden Notebooks bleiben dabei erhalten — es werden nur neue bzw. upstream geänderte Dateien nachgezogen.

## Vor dem ersten Versuch

1. Öffnet **`00_umgebungstest.ipynb`** und führt alle Zellen aus, um zu prüfen, dass alle benötigten Pakete verfügbar sind.
2. Legt für jeden Versuch eine Kopie des zugehörigen Blanko-Notebooks in eurem privaten Arbeitsbereich an (Rechtsklick → *Duplicate*), damit eure Bearbeitung nicht mit einem künftigen Update des Original-Notebooks kollidiert.
