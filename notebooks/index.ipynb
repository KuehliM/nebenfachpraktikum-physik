1. **Dateibrowser:** der zentrale Ort für Ihre Dateien und Ordner. Ordner, Notebooks und Messdaten öffnen Sie per Doppelklick. Über einen Rechtsklick können Sie Dateien kopieren (*Duplicate*), umbenennen (*Rename*), herunterladen (*Download*) und löschen (*Delete*).
2. **Werkzeugleiste des Dateibrowsers:** Ordner anlegen, Dateien hochladen (Pfeil nach oben) und die Dateiliste aktualisieren.
3. **Arbeitsbereich:** Geöffnete Dateien erscheinen hier als Tabs. Schließen Sie alle Tabs, erscheint stattdessen der *Launcher*. Den brauchen Sie im Praktikum nicht.
4. **Menüleiste:** Besonders relevant sind *File* (Dateien speichern und herunterladen), *Edit* (Zellen kopieren, einfügen oder löschen), *Run* (einzelne oder alle Zellen ausführen) und *Kernel* (Kernel neu starten, auch mit anschließendem Ausführen aller Zellen).

### Mit einer Kopie arbeiten

Arbeiten Sie immer in einer eigenen Kopie eines Notebooks. Bearbeiten Sie das Original, übernimmt der automatische Abgleich spätere Korrekturen nur noch teilweise. Ihre Kopie bleibt beim Abgleich dagegen unberührt, und das unveränderte Original erhält alle Korrekturen.

Brauchen Sie ein Notebook noch einmal im Ausgangszustand, legen Sie einfach eine neue Kopie vom Original an.

### Ein geöffnetes Notebook



1. **Dateibrowser:** zeigt den aktuellen Ordner, das geöffnete Notebook ist markiert. Messdaten und Bilder, die Sie im Notebook verwenden, gehören in denselben Ordner wie das Notebook. Dann genügt beim Einlesen der Dateiname.
2. **Notebook-Tab und Werkzeugleiste:** Speichern, Zelle einfügen, Zellen ausschneiden/kopieren/einfügen, Zelle ausführen (▶), Ausführung stoppen (■), Kernel neu starten (↻), Kernel neu starten und alle Zellen ausführen (⏩) sowie die Auswahl des Zelltyps (*Code* oder *Markdown*).
3. **Kernel:** das Programm, das Ihren Code ausführt und alle Variablen im Arbeitsspeicher hält (hier *Python 3.12*). Der Kreis daneben ist leer, wenn der Kernel bereit ist, und gefüllt, während er rechnet. Wird der Kernel neu gestartet, gehen alle Variablen verloren.
4. **Code-Zelle:** enthält Python-Code. Die Zahl links, z. B. `[4]`, zeigt, als wievielte Zelle sie ausgeführt wurde.
5. **Markdown-Zelle:** enthält Text, Erklärungen, Aufgabenstellungen und Formeln. Zum Bearbeiten doppelklicken, mit `Shift` + `Enter` wieder formatiert anzeigen.
6. **Ausgabe:** Ergebnisse, Diagramme und Fehlermeldungen erscheinen direkt unter der Zelle.

### Reihenfolge und Kernel

- Zellen bauen aufeinander auf: Eine Variable existiert erst, wenn die Zelle, in der sie definiert wird, ausgeführt wurde. Die Fehlermeldung `NameError: name '…' is not defined` bedeutet meist, dass eine Zelle weiter oben noch nicht ausgeführt wurde.
- Entscheidend ist die Reihenfolge, in der Sie die Zellen ausführen, nicht ihre Position im Notebook. Führen Sie Zellen deshalb von oben nach unten aus.
- Nach einer längeren Pause oder einem Neustart sind alle Variablen gelöscht. Mit *Run → Run All Cells* stellen Sie den Stand wieder her.
- Steht links `[*]` und es passiert nichts mehr, stoppen Sie die Ausführung (■) oder starten Sie den Kernel neu (↻).

### Wichtige Tastenkürzel

Die Kürzel mit Einzelbuchstaben funktionieren nur, wenn Sie gerade keine Zelle bearbeiten. Drücken Sie dazu vorher `Esc`.

| Aktion | Tastenkürzel |
|--------|--------------|
| Zelle ausführen und zur nächsten springen | `Shift` + `Enter` |
| Zelle bearbeiten / Bearbeiten beenden | `Enter` / `Esc` |
| Neue Zelle darüber / darunter | `A` / `B` |
| Zelle in Markdown / Code umwandeln | `M` / `Y` |
| Zelle löschen / Löschen rückgängig machen | `D` `D` / `Z` |
| Speichern | `Strg` + `S` (Mac: `Cmd` + `S`) |

### Text und Formeln in Markdown-Zellen

Ergebnisse und Diskussionen schreiben Sie in Markdown-Zellen. Formeln stehen in LaTeX-Schreibweise zwischen Dollarzeichen, abgesetzte Formeln zwischen `$$` … `$$`.

| Eingabe | Darstellung |
|---------|-------------|
| `**fett**`, `*kursiv*` | **fett**, *kursiv* |
| `$u_x$`, `$\bar x$`, `$x^2$` | $u_x$, $\bar x$, $x^2$ |
| `$\frac{a}{b}$`, `$\sqrt{x}$` | $\frac{a}{b}$, $\sqrt{x}$ |
| `$(72{,}5 \pm 0{,}8)\,\mathrm{mN/m}$` | $(72{,}5 \pm 0{,}8)\,\mathrm{mN/m}$ |

`{,}` setzt ein Dezimalkomma ohne Leerzeichen, `\,` einen kleinen Abstand vor der Einheit.
