```python
from scipy.optimize import curve_fit

def linear(x, a, b):
    return a * x + b

x = np.array([1, 2, 3, 4, 5])
y = np.array([2.1, 4.1, 6.0, 8.2, 10.1])

popt, pcov = curve_fit(linear, x, y)

print("a =", popt[0])
print("b =", popt[1])
```
