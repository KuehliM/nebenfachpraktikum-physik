## 2. Phasenkurve

In diesem Teil wird aus der gemessenen Zeitdifferenz $\Delta t$ zwischen zwei phasengleichen Punkten von $U_1$ und $U_2$ die Phasenverschiebung $\varphi$ zwischen Spannung und Strom bestimmt (Aufgabe 2):

$$
\varphi = \pm\, 360° \cdot \Delta t \cdot f
$$

Das Vorzeichen folgt der Konvention aus Gl. 8.1: $\varphi > 0$ heißt, dass der Strom der Spannung nacheilt. Unterhalb der Resonanz eilt der Strom voraus ($\varphi < 0$), oberhalb eilt er nach ($\varphi > 0$), bei $f_0$ ist $\varphi = 0$.

Erwartet wird der Verlauf nach Gl. 8.10:

$$
\tan\varphi = \frac{\omega L - \frac{1}{\omega C}}{R_\mathrm{ges}}
$$

Zuerst werden die Messwerte definiert. Danach wird $\varphi$ mit Unsicherheit berechnet und die Phasenkurve gegen $f/f_0$ dargestellt.

### Messwerte definieren

Hier werden die Frequenzen $f$ und die gemessenen Zeitdifferenzen $\Delta t$ definiert. $\Delta t$ wird von $\mathrm{ms}$ in $\mathrm{s}$ umgerechnet.

**Wichtig:** Das Vorzeichen von $\Delta t$ ergibt sich daraus, welches Signal seinen Nulldurchgang zuerst erreicht (Abschnitt 8.4). Da $U_2$ am Widerstand in Phase mit dem Strom ist, eilt der Strom voraus, wenn $U_2$ vor $U_1$ kommt, und $\Delta t$ wird negativ. Kommt $U_2$ später, eilt der Strom nach und $\Delta t$ ist positiv.
