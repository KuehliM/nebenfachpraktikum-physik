### Aufgabe 1a: Planck'sches Wirkungsquantum durch linearen Fit bestimmen

Die Ausgleichsgerade hat die Form:

$$
E_\mathrm{Ph} = m\,f + b
$$

Aus dem Vergleich mit Gl. 5.2 folgt:

$$
h = m
$$

Der Fit liefert außerdem den Achsenabschnitt $b$. Er wird in Aufgabe 1c gebraucht.

Weil beide Größen Unsicherheiten tragen, wird der Fit mit `odr_fit` durchgeführt. Er gewichtet die Messwerte mit ihren Unsicherheiten in beide Richtungen, genau wie die Methode minimaler und maximaler Steigung aus Abschnitt 1.3.2 die Fehlerbalken in $x$- und $y$-Richtung verwendet. Aufgabe 1b verlangt diese Abschätzung zusätzlich, sie folgt im nächsten Schritt.

Zur Erinnerung: ODR berücksichtigt Unsicherheiten in $x$- und $y$-Richtung. Der Fit läuft in vier Schritten.

**1. Modellfunktion definieren.** Die Parameter stehen gemeinsam im Array `beta`. Die Funktion bekommt zuerst `x` und danach `beta`:

```python
def linear_odr(x, beta):
    return beta[0] * x + beta[1]
```

**2. Gewichte berechnen.** `odr_fit` erwartet die Unsicherheiten nicht direkt, sondern als Gewichte. Das Gewicht eines Messwerts ist eins geteilt durch das Quadrat seiner Unsicherheit:

$$
w_x = \frac{1}{u_x^2}
\qquad
w_y = \frac{1}{u_y^2}
$$

**3. Fit durchführen.** Übergeben werden die Modellfunktion, die Messwerte, die Startwerte `beta0` und die Gewichte:

```python
output = odr_fit(linear_odr, x, y, beta0=[1, 0], weight_x=1/u_x**2, weight_y=1/u_y**2)
```

**4. Ergebnisse auslesen.** Die Parameter stehen in `output.beta`. Ihre Unsicherheiten stehen wie bei `curve_fit` auf der Diagonalen der Kovarianzmatrix, hier heißt sie `output.cov_beta`:

```python
u_m = np.sqrt(output.cov_beta[0, 0])   # Steigung
u_b = np.sqrt(output.cov_beta[1, 1])   # Achsenabschnitt
```

**Wichtig:** `beta0` sind die Startwerte des Fits. `odr_fit` arbeitet mit relativen Schrittweiten und braucht sie deshalb in der richtigen Größenordnung. Bei Zahlen wie $10^{-34}$ darf kein Startwert genau null sein, sonst bricht der Fit ab.

### Aufgabe 1b: Unsicherheit von $h$ nach der Methode minimaler und maximaler Steigung bestimmen

Die Anleitung verlangt, die Ausgleichsgerade im Rahmen der Fehlerbalken zu variieren. Dazu werden zwei Grenzgeraden durch den Messpunkt mit der kleinsten und den mit der größten Frequenz gelegt (Methode aus Abschnitt 1.3.2 der Anleitung, Indizes mit `np.argmin()` und `np.argmax()`):

- minimale Steigung durch $(f_\mathrm{min} - u_f,\ E + u_E)$ und $(f_\mathrm{max} + u_f,\ E - u_E)$,
- maximale Steigung durch $(f_\mathrm{min} + u_f,\ E - u_E)$ und $(f_\mathrm{max} - u_f,\ E + u_E)$.

Die Steigung einer Geraden durch zwei Punkte ist der Quotient aus $y$- und $x$-Differenz. Die größere Abweichung zur Steigung des Fits ist die Unsicherheit:

$$
u_h = \max\left( |h - m_\mathrm{max}|,\ |h - m_\mathrm{min}| \right)
$$
