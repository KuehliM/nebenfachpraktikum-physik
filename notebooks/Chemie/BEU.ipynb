## 1. Beugung am Gitter

In diesem Teil wird aus den gemessenen Beugungswinkeln $\theta$ die Gitterkonstante $g$ bestimmt.

Es gilt die Gitterbeziehung (Gl. 3.3):

$$
g \sin\theta = m\lambda
$$

mit der Beugungsordnung $m$ und der Wellenlänge $\lambda$.

Mit $x = m\lambda$ und $y = \sin\theta$ wird daraus eine lineare Funktion:

$$
y = \frac{1}{g}\,x
$$

Die Steigung der Ausgleichsgeraden ist also $1/g$.

Da die Winkelskala des Spektrometers willkürlich orientiert ist, wird jeder Beugungswinkel aus den Skalenwerten rechts $\theta_r$ und links $\theta_l$ vom Maximum nullter Ordnung bestimmt (Nullpunktkorrektur, Gl. 3.5):

$$
\theta = \frac{|\theta_r - \theta_l|}{2}
$$

Zuerst wird das gröbere Gitter mit der Natriumlampe ausgewertet, danach das feinere Gitter mit der Quecksilberlampe.

### Aufgabe 1b: Linearen Fit durchführen (Natrium)

Die Ausgleichsgerade hat die Form:

$$
y = mx + b
$$

Aus dem Vergleich mit $\sin\theta = \dfrac{1}{g}\,m\lambda$ folgt für die Steigung:

$$
\frac{1}{g} = m
$$

Der Achsenabschnitt $b$ sollte nach der Nullpunktkorrektur mit null verträglich sein.

Der Fit wird mit `curve_fit()` durchgeführt. Die Unsicherheiten von $\sin\theta$ werden über `sigma=u_y` berücksichtigt (gewichtete Regression).

Zur Erinnerung: `fit, cov = curve_fit(linear_function, x, y, sigma=u_y, absolute_sigma=True)` mit den $x$-Werten, den $y$-Werten und deren Unsicherheiten. Die Fitfunktion `linear_function(x, m, b)` wird vorher mit `def` definiert.

`curve_fit()` gibt die Fitparameter `fit` und die Kovarianzmatrix `cov` zurück:

$$
\mathrm{cov} =
\begin{pmatrix}
\mathrm{cov}[0,0] & \mathrm{cov}[0,1] \\
\mathrm{cov}[1,0] & \mathrm{cov}[1,1]
\end{pmatrix}
=
\begin{pmatrix}
u_m^2 & \cdots \\
\cdots & u_b^2
\end{pmatrix}
$$

Auf der Diagonalen stehen die quadrierten Unsicherheiten der Fitparameter:

$$
u_m = \sqrt{\mathrm{cov}[0,0]}
\qquad
u_b = \sqrt{\mathrm{cov}[1,1]}
$$

## 2. Brechungsindex von Glas (Michelson-Interferometer)

$$
n = 1 + \frac{Z\lambda\cos\alpha}{2t\,(1-\cos\alpha) - Z\lambda}
$$
