### Aufgabe 2b: Mittelwerte und Unsicherheiten der Mittelwerte berechnen

Für die Viskosität werden die Mittelwerte von Fallzeit, Durchmesser und Masse verwendet.

Die Mittelwerte werden mit `np.mean()` berechnet.

Zusätzlich wird für jede Messreihe die Unsicherheit des Mittelwerts über

$$
u_{\bar x} =
\frac{s_x}{\sqrt{N}}
$$

bestimmt.

$N$ ist die Anzahl der Messwerte. In Python kann die Anzahl der Werte eines Arrays `x` mit `len(x)` bestimmt werden.

$s_x$ ist die Stichprobenstandardabweichung der Messreihe:

$$
s_x =
\sqrt{
\frac{1}{N-1}
\sum_{i=1}^{N}
(x_i-\bar x)^2
}
$$

Die Standardabweichung eines Arrays `x` kann mit `np.std(x)` berechnet werden. Standardmäßig verwendet `np.std()` dabei den Nenner $N$. Für die Stichprobenstandardabweichung soll jedoch der Nenner $N-1$ verwendet werden. Deshalb muss bei `np.std()` der Parameter `ddof=1` gesetzt werden.

Die Unsicherheit des Mittelwerts wird quadratisch mit der Unsicherheit des Messgeräts kombiniert (sie bleibt auch dann bestehen, wenn alle Messwerte gleich sind):

$$
u_x =
\sqrt{
u_{\bar x}^2 + u_{x,\mathrm{Gerät}}^2
}
$$

Für die Berechnung werden also benötigt:

- der Mittelwert mit `np.mean()`,
- die Standardabweichung mit `np.std(..., ddof=1)`,
- die Anzahl der Messwerte mit `len()`,

Die Standardabweichung selbst muss hier nicht als eigenes Ergebnis angegeben werden. Sie wird nur verwendet, um die Unsicherheit des Mittelwerts zu bestimmen.

### Aufgabe 3b und 3c: Linearen Fit mit `curve_fit` durchführen

Die linearisierte Funktion hat die Form:

$$
y = mx+b
$$

Aus dem Vergleich mit

$$

$$

$$

Der Fit wird mit `curve_fit()` durchgeführt. Die Unsicherheiten von $\ln(\eta)$ werden dabei über `sigma=u_y` berücksichtigt (gewichtete Regression).

Zur Erinnerung: `fit, cov = curve_fit(linear_function, x, y, sigma=u_y, absolute_sigma=True)` mit den $x$-Werten, den $y$-Werten und deren Unsicherheiten. Die Fitfunktion `linear_function(x, m, b)` wird vorher mit `def` definiert.

`curve_fit()` gibt die Fitparameter `fit` und die Kovarianzmatrix `cov` zurück:

$$
\mathrm{cov} =
\begin{pmatrix}
\mathrm{cov}[0,0] & \mathrm{cov}[0,1] \\
\mathrm{cov}[1,0] & \mathrm{cov}[1,1]
\end{pmatrix}
=
\begin{pmatrix}
u_m^2 & \cdots \\
\cdots & u_b^2
\end{pmatrix}
$$

Auf der Diagonalen stehen die quadrierten Unsicherheiten der Fitparameter:

$$
u_m = \sqrt{\mathrm{cov}[0,0]}
\qquad
u_b = \sqrt{\mathrm{cov}[1,1]}
$$

Die Unsicherheit von $\alpha = e^b$ folgt aus Fehlerfortpflanzung. Wegen

$$
\frac{\partial \alpha}{\partial b} = e^b = \alpha
$$

gilt:

$$
u_\alpha = \left|\frac{\partial \alpha}{\partial b}\right| u_b = \alpha\, u_b
$$

## Abgabe als Versuchsbericht

Wenn Sie dieses Notebook als Versuchsbericht abgeben, prüfen Sie vorher, dass alle Aufgaben ausgewertet und Ihre Ergebnisse in Markdown-Zellen diskutiert sind, und füllen Sie die Kopfzelle aus. Laden Sie das Notebook als `.ipynb`-Datei bis zum nächsten Termin im moodle-Lernraum hoch. Gehören weitere Dateien dazu, zum Beispiel eine CSV-Datei, aus der Sie Messdaten eingelesen haben, oder Bilder, die nicht im Notebook eingebettet sind, laden Sie das Notebook zusammen mit diesen Dateien als `.zip`-Archiv hoch. Benennen Sie die Datei nach dem Schema `Gruppe_Nachname1_Nachname2_VIS` mit der Bezeichnung Ihrer Gruppe (zum Beispiel CMo03) und den Nachnamen aller Gruppenmitglieder. Bei einer Kurzveröffentlichung entfällt die Abgabe des Notebooks (Hinweis am Anfang).
