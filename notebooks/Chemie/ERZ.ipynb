### Aufgabe 1a: Mittelwert und Unsicherheit der Periodendauer berechnen

Für jede Messreihe wird der Mittelwert $\bar T$ mit `np.mean()` berechnet.

Zusätzlich wird die Unsicherheit des Mittelwerts über

$$
u_{\bar T} =
\frac{s_T}{\sqrt{N}}
$$

bestimmt. $N$ ist die Anzahl der Messungen, `len(T)` liefert $N$.

$s_T$ ist die Stichprobenstandardabweichung der Messreihe:

$$
s_T =
\sqrt{
\frac{1}{N-1}
\sum_{i=1}^{N}
(T_i-\bar T)^2
}
$$

`np.std(T)` verwendet standardmäßig den Nenner $N$. Für die Stichprobenstandardabweichung muss deshalb `ddof=1` gesetzt werden: `np.std(T, ddof=1)`.

Die Unsicherheit des Mittelwerts wird quadratisch mit der Ableseunsicherheit kombiniert (sie bleibt auch dann bestehen, wenn alle Messwerte gleich sind):

$$
u_T =
\sqrt{
u_{\bar T}^2 + u_{T,\mathrm{Ablese}}^2
}
$$

Zur Erinnerung: `fit, cov = curve_fit(linear_function, x, y, sigma=u_y, absolute_sigma=True)` mit den $x$-Werten, den $y$-Werten und deren Unsicherheiten. Die Fitfunktion `linear_function(x, m, b)` wird vorher mit `def` definiert.

Der Fit gewichtet die Messwerte mit ihren Unsicherheiten. Die Unsicherheiten der Fitparameter stehen in der Kovarianzmatrix, sie ersetzen hier die Methode minimaler und maximaler Steigung aus Abschnitt 1.3.2.

`curve_fit()` gibt die Fitparameter `fit` und die Kovarianzmatrix `cov` zurück:

$$
\mathrm{cov} =
\begin{pmatrix}
\mathrm{cov}[0,0] & \mathrm{cov}[0,1] \\
\mathrm{cov}[1,0] & \mathrm{cov}[1,1]
\end{pmatrix}
=
\begin{pmatrix}
u_m^2 & \cdots \\
\cdots & u_b^2
\end{pmatrix}
$$

Auf der Diagonalen stehen die quadrierten Unsicherheiten der Fitparameter:

$$
u_m = \sqrt{\mathrm{cov}[0,0]}
\qquad
u_b = \sqrt{\mathrm{cov}[1,1]}
$$
